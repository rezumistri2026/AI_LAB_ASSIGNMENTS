```python
!pip install -q kagglehub[pandas-datasets]

import kagglehub, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display

# Download dataset
path = kagglehub.dataset_download(
    "kristoft/pitt-quantum-repository-106066-molecules"
)

# Find and load JSON file
file = next(
    os.path.join(r, f)
    for r, _, files in os.walk(path)
    for f in files if f.endswith(".json")
)
df = pd.read_json(file)

# Dataset overview
print("First 5 Records:\n", df.head())
print("\nLast 5 Records:\n", df.tail())
print("\nShape:", df.shape)
print("\nColumns:", list(df.columns))
df.info()

# Statistics and data cleaning
display(df.describe(include="all"))
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)
print("\nUpdated Shape:", df.shape)

# Column types
print("\nNumerical:", df.select_dtypes(include="number").columns.tolist())
print("\nCategorical:", df.select_dtypes(include="object").columns.tolist())

# Histograms
for col in df.select_dtypes(include="number").columns[:5]:
    df[col].hist(bins=30, figsize=(6, 4))
    plt.title(col)
    plt.show()

# Correlation heatmap
num = df.select_dtypes(include="number")
if num.shape[1] > 1:
    sns.heatmap(num.corr(), annot=True, cmap="coolwarm")
    plt.title("Correlation Heatmap")
    plt.show()

# Final summary
print("\nFinal Dataset Summary")
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Missing Values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```
